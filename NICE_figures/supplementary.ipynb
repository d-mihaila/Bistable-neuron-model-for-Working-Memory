# Supplementary figures (not in the paper)

1. **Robustness** (Section III, numbers in the text only): accuracy against delay, weight noise and input noise, MNIST and N-MNIST. Weights: every synapse scaled by $(1+\mathrm{CV}\,\xi)$, $\xi\sim\mathcal{N}(0,1)$. Input: each pixel rate (MNIST) or each channel's event count (N-MNIST) scaled the same way. Mean over 5 noise draws, read out at the trained 100 ms delay. Scripts: `exp_noise_weights_cv.py`, `exp_noise_input_cv.py`, `run_noise_cv.sh` in `../MNIST/` and `../N_MNIST/`.
2. **Rasters** of every MNIST network over the full 5 s.
3. **Learnt $w_0$** of the two-tailed network, before and after training.
4. **The ODR network**: ring schematic and Mexican-hat E → E weights.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import style as S
S.apply()
import mnist_figures as M
import odr_figures as O

data = M.load("MNIST")
data_nmnist = M.load("N-MNIST")
```

```python
fig = M.make_figure_accuracy(data=data)
S.save(fig, "supp_mnist_delay_noise")
plt.show()
fig = M.make_figure_accuracy(data=data_nmnist)
S.save(fig, "supp_nmnist_delay_noise")
plt.show()
```

Accuracy (%) without noise and at a CV of 50 %, and the loss in points.

```python
def noise_table(d, key):
    runs = M.runs_present(d)
    df = d[key][d[key].run_name.isin(runs)]
    m = df.groupby(["run_name", "cv"]).test_acc.mean().unstack() * 100
    out = pd.DataFrame({"CV 0": m[0.0], "CV 50 %": m[0.5]}).reindex(runs)
    out["loss (points)"] = out["CV 0"] - out["CV 50 %"]
    out.index = [S.net(r)["label"] for r in runs]
    return out.round(1)

for name, d in (("MNIST", data), ("N-MNIST", data_nmnist)):
    for key in ("weight_cv", "input_cv"):
        print(name, key)
        display(noise_table(d, key))
```
